# INRTODUCTION/OVERVIEW

```python
import pandas as pd
import ast
from sklearn.feature_extraction.text import TfidfVectorizer as tfv
from sklearn.metrics.pairwise import cosine_similarity as cs
import joblib
```

```python
df = pd.read_csv("tmdbcredits.csv", engine='python', on_bad_lines='skip')
dff = pd.read_csv("tmdbmovies.csv")
dff.head()
```

```python
df.tail()
```

```python
dff.columns
```

```python
df.columns
```

```python

dff["release_date"] = pd.to_datetime(dff["release_date"])
dff.info()
```

```python
tag = dff[["title","genres","overview","keywords","tagline"]]
tag.tail()
```

```python
df = df["cast"]
df.head()
```

# FEATURE SELECTION

```python
tag = pd.concat([tag,df],axis=1)
tag.tail()
```

# Columns like "genres", "keywords" and "cast" are in JSON-like structure we need to extrect them to normal/plain text

```python
tag.dropna(inplace=True)
tag.reset_index(drop=True, inplace=True)
tag.isnull().sum()
```

# JSON Extraction

```python
def extract_names(text):

    names = []

    for item in ast.literal_eval(text):
        names.append(item["name"])

    return " ".join(names)
```

```python
tag["genres"] = tag["genres"].apply(extract_names)
tag["keywords"] = tag["keywords"].apply(extract_names)
tag["cast"] = tag["cast"].apply(extract_names)
```

# Now we have just converted JSON-like structure to plain text with above code ready for Vectorization

```python
tag.head()
```

# So now how do i measure similarity between Movies? By converting text to numbers(vectors) using Vectorization

# Our exact work here is called "Content-based filtering"

Now before Vectoriation we need to combine all useful features into ONE (FEATURE ENGINEERING)

Why because TF-IDF works best on a single text field.

# FEATURE ENGINEERING and TEXT COMBINATION

```python
tag["tags"] = (
   tag["cast"] + " " +
    tag["genres"] + " " +
    tag["overview"] + " " +
    tag["keywords"]
)
tag.tail()
```

# TF-IDF Vectorization

```python
vector = tfv(stop_words="english")
vector_matrix = vector.fit_transform(tag["tags"])
```

# COSINE SIMILARITY

```python
similarity = cs(vector_matrix)
similarity
```

# RECOMMENDATION FUNCTION

```python
def recommend(movie_name):

    movie_index = tag[tag["title"] == movie_name].index[0]

    distances = similarity[movie_index]

    movie_list = sorted(
        list(enumerate(distances)),
        reverse=True,
        key=lambda x: x[1]
    )[1:6]  # top 5 (excluding itself)

    for i in movie_list:
        print(tag.iloc[i[0]].title)
```

# TESTING RECOMMENDTIONS

```python
recommend("Avatar")
```

```python
joblib.dump(tag, "movies.pkl")
joblib.dump(similarity, "similarity.pkl")
```

# FINDINGS
- TF-IDF combined some important features together because vectorizer can only use one text column rather than two/more
- Cosine Similarity is then used to find similarities between movies categories mathematically before the creation of "recommendation function" that gives us the actual recommendation we need after inputting any movie name.
